## Cell 2 · Sub-question 1: Fit

First, `has_genre` checks one game's `genres` text, like `"Action, Indie"`. It returns `True` if the student's `GENRE` is in it. `str()` turns a missing genre into the text `"nan"`, so `in` still works (it just won't match). `.apply` sends it one value at a time, the same way as `temp_band` in Week 11, so it reads `GENRE` from Cell 1.

Then `find_fits` works like a funnel. It starts with all 1,000 games and keeps only the ones that:

1. have at least 85% positive reviews,
2. list the student's genre (`.apply(has_genre)` gives True or False for each row),
3. cost no more than the student's price limit, and
4. run on the student's laptop.

It returns the whole table of games that fit, because Sub-questions 2 and 3 need its other columns (like `appid` and `owners_low`).

```python
def has_genre(genres):
    return GENRE in str(genres)

def find_fits(games, max_price, system):
    fits = games[games["pct_pos_total"] >= 85]
    fits = fits[fits["genres"].apply(has_genre)]
    fits = fits[fits["price"] <= max_price]
    fits = fits[fits[system] == True]
    return fits

fits = find_fits(games, MAX_PRICE, SYSTEM)
print(len(fits), "games fit the student's choices")
print(fits[["name", "price", "pct_pos_total"]].head(10))
```

## Cell 3 · Asking CheapShark about one game

These are the same CheapShark requests as Module 5, Exercises 9 to 11. Every request sends `headers=HEADERS`, because CheapShark blocks requests that don't say what program sent them.

**`get_today(appid)`** returns a game's cheapest price in any store today:

1. `time.sleep(1)` waits 1 second, so we don't send requests too fast.
2. It asks with `{"steamAppID": appid}`. If the status code isn't 200, the request failed, so it returns `0`.
3. `.json()` gives a list with one dictionary. If the list is empty, CheapShark doesn't sell the game, so it returns `0`.
4. Otherwise it returns `"cheapest"` from `data[0]`, turned from text into a number with `float()`.

**`get_lowest_ever(appid)`** is Exercise 11: the same first request, then a second request with `{"id": game_id}` (CheapShark's own ID), which gives the lowest price ever under `"cheapestPriceEver"`, then `"price"`.

Both return `0` when CheapShark has no answer, so one missing game can't break the app. Cell 5 then leaves those games out.

```python
URL = "https://www.cheapshark.com/api/1.0/games"
HEADERS = {"User-Agent": "MIST4600-dorm-room-deals"}

def get_today(appid):
    time.sleep(1)
    response = requests.get(URL, params={"steamAppID": appid}, headers=HEADERS)
    if response.status_code != 200:
        return 0
    data = response.json()
    if len(data) == 0:
        return 0
    return float(data[0]["cheapest"])

def get_lowest_ever(appid):
    time.sleep(1)
    response = requests.get(URL, params={"steamAppID": appid}, headers=HEADERS)
    if response.status_code != 200:
        return 0
    data = response.json()
    if len(data) == 0:
        return 0
    game_id = data[0]["gameID"]
    response2 = requests.get(URL, params={"id": game_id}, headers=HEADERS)
    if response2.status_code != 200:
        return 0
    data2 = response2.json()
    return float(data2["cheapestPriceEver"]["price"])
```

Try them on one game, Stardew Valley (`appid` 413150). You should see two prices: today's and the lowest ever.

```python
print(get_today(413150))
print(get_lowest_ever(413150))
```

## Cell 4 · A class for each pick

Each game checked with CheapShark becomes a `DealPick` object with its name, its regular price (from `steam_games.csv`), its price today, and its lowest price ever.

* `on_sale()` returns `True` if today's price is lower than the regular price.
* `card()` returns one line of text for the student, built with an f-string.

```python
class DealPick:
    def __init__(self, name, regular_price, today_price, lowest_price):
        self.name = name
        self.regular_price = regular_price
        self.today_price = today_price
        self.lowest_price = lowest_price

    def on_sale(self):
        return self.today_price < self.regular_price

    def card(self):
        return f"{self.name}: ${self.today_price} today, regular ${self.regular_price}, lowest ever ${self.lowest_price}"
```

Try the class with a made-up pick:

```python
test = DealPick("Test Game", 14.99, 7.49, 5.99)
print(test.card())
print(test.on_sale())
```

## Cell 5 · Sub-question 2: Deals

`find_deals` works like Module 5, Exercise 12:

1. It keeps the games that fit and cost money (free games are never on sale) and takes the 5 with the best reviews. Only 5 games are sent to CheapShark, not all of them. `.copy()` lets us add columns to this smaller table without a pandas warning (Week 12).
2. `.apply(get_today)` and `.apply(get_lowest_ever)` add two new columns: today's price and the lowest price ever, one request per game.
3. It keeps only the games CheapShark found (`today` above 0).
4. `list(...)` turns each column into a plain list. The loop goes through the positions `0, 1, 2, ...` and makes one `DealPick` per game, from the same position in each list.

Then we print each pick's card (today's price next to the regular price and the lowest price ever), and the picks that are on sale today.

```python
def find_deals(fits):
    paid = fits[fits["price"] > 0]
    shortlist = paid.sort_values("pct_pos_total", ascending=False).head(5).copy()
    shortlist["today"] = shortlist["appid"].apply(get_today)
    shortlist["lowest"] = shortlist["appid"].apply(get_lowest_ever)
    shortlist = shortlist[shortlist["today"] > 0]

    names = list(shortlist["name"])
    regular = list(shortlist["price"])
    today = list(shortlist["today"])
    lowest = list(shortlist["lowest"])

    picks = []
    for i in range(len(names)):
        picks.append(DealPick(names[i], regular[i], today[i], lowest[i]))
    return picks

picks = find_deals(fits)

for pick in picks:
    print(pick.card())

print("On sale today:")
for pick in picks:
    if pick.on_sale():
        print("-", pick.name)
```

## Cell 6 · Sub-question 3: Hidden gems

`compare_hidden_gems` splits the games that fit into two groups by their number of owners (`owners_low`):

* **hidden gems:** under 1 million owners
* **big hits:** 1 million owners or more

It first drops games where `owners_low` is 0, because 0 means the owner count is missing, not that nobody owns the game. Then it prints how many games are in each group and their average percent of positive reviews: are the lesser-known games rated as highly as the big hits? It returns the hidden gems, and we print the 5 best-reviewed ones.

```python
def compare_hidden_gems(fits):
    known = fits[fits["owners_low"] > 0]
    gems = known[known["owners_low"] < 1000000]
    hits = known[known["owners_low"] >= 1000000]
    print("Hidden gems:", len(gems), "games, average rating", round(gems["pct_pos_total"].mean(), 1))
    print("Big hits:", len(hits), "games, average rating", round(hits["pct_pos_total"].mean(), 1))
    return gems

gems = compare_hidden_gems(fits)
best_gems = gems.sort_values("pct_pos_total", ascending=False).head(5)
print(best_gems[["name", "price", "pct_pos_total"]])
```
